```python
class Person:
    def __init__(self, ob):
        self.name = ob['name']
        self.last_name = ob['last_name']
        self.company = ob['company']
    
    def full_name(self):
        return self.name + ' ' + self.last_name
    
    def get_email(self):
        return self.name + '.' + self.last_name + '@' + self.company + '.co.za'
```

```python
a = {
    "name":"Azabenathi",
    "last_name":"Pupuma",
    "company":"slipstreamdata"
}
```

```python
full_name = Person(a).full_name()
email = Person(a).get_email()

print(full_name)
print(email)
```
