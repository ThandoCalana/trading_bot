```python
import requests
import pandas as pd
import defs
```

```python
session = requests.Session()
```

```python
instrument = 'EUR_USD'
count = 10
granularity = 'H1'
```

```python
url = f'{defs.OANDA_URL}/instruments/{instrument}/candles'
```

```python
url
```

```python
params = dict(
    count = count,
    granularity = granularity,
    price = 'MBA'
)
```

```python
params
```

```python
response = session.get(url, params=params, headers=defs.SECURE_HEADER)
```

```python
response.status_code
```

```python
data = response.json()
data
```

```python
data.keys()
```

```python
len(data['candles'])
```

```python
prices = ['mid', 'bid', 'ask']
ohlc = ['o', 'h', 'l', 'c']
```

```python
our_data = []
for candle in data['candles']:
    if not candle['complete']:
        continue
    new_dict = {}
    new_dict['time'] = candle['time']
    new_dict['volume'] = candle['volume']
    for price in prices: 
        for o in ohlc:
            new_dict[f'{price}_{o}'] = candle[price][o]
    our_data.append(new_dict)
```

```python
our_data[0]
```

```python
candles_df = pd.DataFrame(our_data)
candles_df
```

```python
candles_df.to_pickle('EURUSD.pkl')
test_df = pd.read_pickle('EURUSD.pkl')
test_df
```
